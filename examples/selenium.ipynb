```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time
import re

# Set up Chrome options
chrome_options = Options()
# Uncomment the line below to run Chrome in headless mode
chrome_options.add_argument("--headless")

# Set up the Selenium WebDriver
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=chrome_options)

# Navigate to the desired URL
driver.get("https://www.allaboutbirds.org/guide/European_Starling/overview")

# Wait for the page to load and perform your scraping logic here
time.sleep(3)
```

```python
print(driver.page_source)
```

```python
# find the audio element
# start by finding the ".hero-content" div
hero_content = driver.find_element(By.CLASS_NAME, "hero-content")
print(hero_content.get_attribute("innerHTML"))
```

```python
# next find the audio tag
audio_tag = hero_content.find_element(By.TAG_NAME, "audio")
print(audio_tag.get_attribute("src"))
```

```python
# now lets find the image
driver.get("https://www.allaboutbirds.org/guide/European_Starling/maps-range")
time.sleep(3)
```

```python
# locate the div.main-area
main_area = driver.find_element(By.CLASS_NAME, "main-area")
# print(main_area.get_attribute("innerHTML"))
# now find the img tag
img_tag = main_area.find_element(By.TAG_NAME, "img")
print(img_tag.get_attribute("src"))
```

```python
# now lets find the <p> tag with id = "migration-description"
migration_description = driver.find_element(By.ID, "migration-description")
print(migration_description.get_attribute("innerHTML"))
```

```python
# now lets get their life history
driver.get("https://www.allaboutbirds.org/guide/European_Starling/lifehistory")
time.sleep(3)
```

```python
# get thier habitat
lifehistory_menu = driver.find_element(By.CLASS_NAME, "LH-menu")
# print(lifehistory_menu.get_attribute("innerHTML"))

habitat = lifehistory_menu.find_element(By.XPATH, ".//li[a[contains(@href, '#habitat')]]")
habitat_text = habitat.find_element(By.CLASS_NAME, "text-label")
span_elements = habitat_text.find_elements(By.TAG_NAME, "span")
if len(span_elements) > 1:
    habitat_value = span_elements[1].get_attribute("innerHTML")
    print(habitat_value)

food = lifehistory_menu.find_element(By.XPATH, ".//li[a[contains(@href, '#food')]]")
food_text = food.find_element(By.CLASS_NAME, "text-label")
span_elements = food_text.find_elements(By.TAG_NAME, "span")
if len(span_elements) > 1:
    food_value = span_elements[1].get_attribute("innerHTML")
    print(food_value)

nesting = lifehistory_menu.find_element(By.XPATH, ".//li[a[contains(@href, '#nesting')]]")
nesting_text = nesting.find_element(By.CLASS_NAME, "text-label")
span_elements = nesting_text.find_elements(By.TAG_NAME, "span")
if len(span_elements) > 1:
    nesting_value = span_elements[1].get_attribute("innerHTML")
    print(nesting_value)

behavior = lifehistory_menu.find_element(By.XPATH, ".//li[a[contains(@href, '#behavior')]]")
behavior_text = behavior.find_element(By.CLASS_NAME, "text-label")
span_elements = behavior_text.find_elements(By.TAG_NAME, "span")
if len(span_elements) > 1:
    behavior_value = span_elements[1].get_attribute("innerHTML")
    print(behavior_value)

conservation = lifehistory_menu.find_element(By.XPATH, ".//li[a[contains(@href, '#conservation')]]")
conservation_text = conservation.find_element(By.CLASS_NAME, "text-label")
span_elements = conservation_text.find_elements(By.TAG_NAME, "span")
if len(span_elements) > 1:
    conservation_value = span_elements[1].get_attribute("innerHTML")
    print(conservation_value)
```

```python
# finally, lets get their description
driver.get("https://www.allaboutbirds.org/guide/European_Starling/overview")
time.sleep(3)
```

```python
# description
overview_div = driver.find_element(By.XPATH, "//div[h2[contains(@class, 'overview')]]")
first_p_tag = overview_div.find_element(By.TAG_NAME, "p")
description = first_p_tag.get_attribute("innerHTML")
print(description)

# backyard tips
tips_div = driver.find_element(By.XPATH, "//div[h2[text()='Backyard Tips']]")
second_p_tag = tips_div.find_elements(By.TAG_NAME, "p")[1]
tips = second_p_tag.get_attribute("innerHTML")
tips = re.sub(r'<a[^>]*>(.*?)<\/a>', '', tips) # remove the links
print(tips)

# find this bird
find_this_bird_div = driver.find_element(By.XPATH, "//div[h2[text()='Find This Bird']]")
second_p_tag = find_this_bird_div.find_elements(By.TAG_NAME, "p")[1]
find_this_bird = second_p_tag.get_attribute("innerHTML")
print(find_this_bird)
```

```python
# Close the browser
driver.quit()
```

```python
# Navigate to the desired URL
driver.get("https://www.allaboutbirds.org/guide/Mottled_Duck/maps-range")
# Wait for the page to load and perform your scraping logic here
time.sleep(1)
```

```python
migration_description_div = driver.find_element(By.XPATH, "//div//p[@id='migration-description']").find_element(By.XPATH, "..")

# Find all <p> tags within the migration description div
p_tags = migration_description_div.find_elements(By.TAG_NAME, "p")

# Initialize variables to track the largest innerHTML and corresponding <p> tag
largest_inner_html = ""

# Iterate through the <p> tags to find the one with the largest innerHTML
for p in p_tags:
    inner_html = p.get_attribute("innerHTML")
    if len(inner_html) > len(largest_inner_html):
        largest_inner_html = inner_html

# Get the innerHTML of the largest <p> tag
migration_description = largest_inner_html


print(migration_description)
```

## Find Unique ebird_extra habitats

```python
db_path = 'BirdNET_UI/ebirds.db'

# load database
import sqlite3

# Connect to the ebirds database
conn = sqlite3.connect(db_path)

# Create a cursor object
cursor = conn.cursor()

# get all unique habitat
# Execute a query to get unique habitats from the database
cursor.execute("SELECT DISTINCT habitat_value FROM ebirds_extra")
unique_habitats = cursor.fetchall()

# Print the unique habitats
for habitat in unique_habitats:
    print(habitat[0])

# give an example record for each habitat
# Create a dictionary to hold the first example record for each habitat
habitat_examples = {}

# Iterate through the unique habitats
for habitat in unique_habitats:
    habitat_name = habitat[0]
    
    # Execute a query to get the first record for the current habitat
    cursor.execute("SELECT * FROM ebirds_extra WHERE habitat_value = ? LIMIT 1", (habitat_name,))
    example_record = cursor.fetchone()
    
    # Store the example record in the dictionary
    if example_record:
        habitat_examples[habitat_name] = example_record

def create_url(common_name):
    filtered_name = common_name.replace(" ", "_").replace("'", "")
    url = f"https://www.allaboutbirds.org/guide/{filtered_name}/overview"
    return url

# Print the first example record for each habitat
for habitat, record in habitat_examples.items():
    print(f"Habitat: {habitat}, Example URL: {create_url(record[2])}")


# Close the connection
conn.close()
```
